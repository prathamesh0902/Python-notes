# Banking App

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def show_details(self):
        return f"Thank you, {self.age} year old, {self.name.title()}"
    
class Bank(User):
    total_deposits =0
    total_withdraws = 0

    def __init__(self, name, age, balance):
        super().__init__(name, age)
        self.balance = balance

    def show_info(self):
        return f"{self.name} has a remaining balance of: ${round(self.balance, 2)}"

    def deposit(self):
        dp = float(input(f"{self.name.title()}, please enter the amount you would like to deposit: "))
        print("Thank you for depositing...")
        self.balance += dp
        Bank.total_deposits += 1                    # Update class variable
        return f"Your balance is now: {round(self.balance,2)}"
    
    def withdraws(self):
        wd = float(input(f"{self.name.title()}, please enter the amount you would like to withdraw: "))
        if wd > self.balance:
            return "Insufficient funds..."
        else:
            print("Thank you for withdrawing...")
            self.balance -= wd
            Bank.total_withdraws += 1                # Update class variable
            return f"Your balance is now: {round(self.balance,2)}"
    
def options(user_bank):
    print("Thank you for creating your bank acconut")
    
    while True:
        print("choose an option :\n1. Show Balance\n2. Deposit\n3. Withdraw\n4. Total Deposits\n5. Total Withdraws\n6. Exit")
        option_choice = int(input("Enter choice :"))
        if option_choice == 1:
            print(user_bank.show_info()) 
        elif option_choice == 2:
            print(user_bank.deposit())
        elif option_choice == 3:
            print(user_bank.withdraws())
        elif option_choice == 4:
            print(f"Total deposits: {Bank.total_deposits}")
        elif option_choice == 5:
            print(f"Total withdraws: {Bank.total_withdraws}")
        elif option_choice == 6:
            print("Thank you for banking with us!")
            return False
            break
        else:
            print("Invalid option, please try again...")


def bank_creation(name):
    balance = float(input(f"{name.title()}, how much money do you have? $"))
    return balance

case = True

while case:
    print("====== Welcome to PCE Bank ======")
    name = input("Enter your name :")
    age = int(input("Enter your age :"))
    user_one = User(name, age)
    user_one_balance = bank_creation(user_one.name)
    user_one_bank = Bank(user_one.name, user_one.age, user_one_balance)
    case = options(user_one_bank)
```

# OOP Concepts

```python
# ============================================================
#  Simple Bank — Beginner OOP in Python
#  Concepts introduced ONE AT A TIME, step by step
# ============================================================
#
#  What you will learn here:
#   Step 1 — What is a Class and Object?
#   Step 2 — The __init__ method (constructor)
#   Step 3 — Instance attributes  (self.something)
#   Step 4 — Instance methods
#   Step 5 — Encapsulation  (hiding data with __)
#   Step 6 — Inheritance    (child class)
#   Step 7 — __str__        (pretty printing)
# ============================================================
```

```python

# ────────────────────────────────────────────────────────────
# STEP 1 & 2 — Your first class + constructor
# ────────────────────────────────────────────────────────────
#
#  Think of a CLASS like a cookie cutter.
#  Think of an OBJECT like the actual cookie made from it.
#
#  class BankAccount:  ← this is the cookie cutter
#  my_account = BankAccount(...)  ← this is ONE cookie (object)
#
#  __init__ is called automatically the moment you create
#  an object. It "initialises" (sets up) the object.

class BankAccount:                          # define the class

    # __init__ runs when you do: BankAccount("Alice", 1000)
    # 'self' refers to the specific object being created
    def __init__(self, owner, balance):
        # STEP 3 — Instance attributes
        # These are stored INSIDE the object using self.
        # Every object gets its own copy of these.
        self.owner   = owner       # who owns this account
        self.balance = balance     # how much money is in it


# ── Try it ───────────────────────────────────────────────────
acc1 = BankAccount("Alice", 1000)   # create object 1
acc2 = BankAccount("Bob",    500)   # create object 2

# Each object has its own data
print(acc1.owner,   acc1.balance)   # Alice  1000
print(acc2.owner,   acc2.balance)   # Bob     500
```

```python
# ────────────────────────────────────────────────────────────
# STEP 4 — Methods
# ────────────────────────────────────────────────────────────
#
#  A METHOD is a function that belongs to a class.
#  It always receives 'self' as its first argument so it
#  can read and change the object's own data.


print("\n" + "─"*45)
print("STEP 4 — Methods (things the object can DO)")
print("─"*45)

class BankAccount:

    def __init__(self, owner, balance):
        self.owner   = owner
        self.balance = balance

    # Method: add money
    def deposit(self, amount):
        self.balance = self.balance + amount          # update balance
        print(f"  Deposited ${amount}. New balance: ${self.balance}")

    # Method: take money out
    def withdraw(self, amount):
        if amount > self.balance:                     # simple check
            print("  Sorry, not enough money!")
        else:
            self.balance = self.balance - amount
            print(f"  Withdrew ${amount}. New balance: ${self.balance}")

    # Method: just show the balance
    def show_balance(self):
        print(f"  {self.owner}'s balance: ${self.balance}")


# ── Try it ───────────────────────────────────────────────────
alice = BankAccount("Alice", 1000)
print("Initial info by using direct variables :",alice.owner,   alice.balance)   # Alice  1000
alice.show_balance()        #  Alice's balance: $1000
alice.deposit(500)          #  Deposited $500.  New balance: $1500
alice.withdraw(200)         #  Withdrew $200.   New balance: $1300
alice.withdraw(9999)        #  Sorry, not enough money!
```

```python
# ────────────────────────────────────────────────────────────
# STEP 5 — Encapsulation
# ────────────────────────────────────────────────────────────
#
#  Problem: right now anyone can do alice.balance = 999999
#  and bypass all your checks. That's dangerous!
#
#  Solution: use double underscore (__balance) to make it
#  PRIVATE. Now it can only be changed through your methods.
#
#  To let people READ the balance safely, you add a
#  get_balance() method — a "getter".


print("\n" + "─"*45)
print("STEP 5 — Encapsulation (hiding private data)")
print("─"*45)

class BankAccount:

    def __init__(self, owner, balance):
        self.owner    = owner
        self.__balance = balance    # __ makes it private!

    # Getter — safe read-only window into the private data
    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount <= 0:
            print("  Deposit must be more than zero.")
            return
        self.__balance += amount    # += means: balance = balance + amount
        print(f"  Deposited ${amount}. Balance: ${self.__balance}")

    def withdraw(self, amount):
        if amount <= 0:
            print("  Withdrawal must be more than zero.")
            return
        if amount > self.__balance:
            print("  Not enough funds!")
            return
        self.__balance -= amount    # -= means: balance = balance - amount
        print(f"  Withdrew ${amount}. Balance: ${self.__balance}")


# ── Try it ───────────────────────────────────────────────────
alice = BankAccount("Alice", 1000)
alice.deposit(500)
alice.withdraw(200)
print(f"  Balance via getter: ${alice.get_balance()}")

# Try to cheat — Python protects it
# alice.__balance = 999999   ← this won't work (AttributeError)
```

```python
# ────────────────────────────────────────────────────────────
# STEP 6 — Inheritance
# ────────────────────────────────────────────────────────────
#
#  Inheritance = a child class GETS everything from the
#  parent class for free, and can ADD its own extras.
#
#  SavingsAccount IS-A BankAccount, but with interest.
#
#  super().__init__(...)  means:
#  "run the parent class's __init__ first, then mine"

print("\n" + "─"*45)
print("STEP 6 — Inheritance (child class)")
print("─"*45)

class BankAccount:                           # PARENT class

    def __init__(self, owner, balance):
        self.owner     = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount <= 0:
            print("  Amount must be positive.")
            return
        self.__balance += amount
        print(f"  Deposited ${amount}. Balance: ${self.__balance}")

    def withdraw(self, amount):
        if amount <= 0:
            print("  Amount must be positive.")
            return
        if amount > self.__balance:
            print("  Not enough funds!")
            return
        self.__balance -= amount
        print(f"  Withdrew ${amount}. Balance: ${self.__balance}")


class SavingsAccount(BankAccount):           # CHILD class

    # (BankAccount) after the name means "inherit from BankAccount"
    def __init__(self, owner, balance, interest_rate):
        super().__init__(owner, balance)     # run parent setup first
        self.interest_rate = interest_rate   # add our own new attribute

    # New method that only SavingsAccount has
    def add_interest(self):
        interest = self.get_balance() * self.interest_rate
        interest = round(interest, 2)        # round to 2 decimal places
        self.deposit(interest)               # reuse deposit() from parent!
        print(f"  Interest earned: ${interest}")


# ── Try it ───────────────────────────────────────────────────
savings = SavingsAccount("Alice", 1000, 0.05)   # 5% interest rate
savings.deposit(500)          # inherited from BankAccount
savings.withdraw(200)         # inherited from BankAccount
savings.add_interest()        # new method only on SavingsAccount
print(f"  Final balance: ${savings.get_balance()}")


```

```python
print("\n" + "─"*45)
print("STEP 7 — Polymorphism (same call, different behaviour)")
print("─"*45)
 
# ────────────────────────────────────────────────────────────
# STEP 7 — Polymorphism
# ────────────────────────────────────────────────────────────
#
#  "Poly" = many.  "Morph" = form.
#  Same method NAME, but each class does it differently.
#
#  Real-life analogy:
#    You say "make a sound" to a Dog  → it barks
#    You say "make a sound" to a Cat  → it meows
#    Same instruction, different result.
#
#  In our bank:
#    You say withdraw(500) to a SavingsAccount  → checks minimum balance
#    You say withdraw(500) to a CurrentAccount  → allows going into overdraft
#    You say withdraw(500) to a BankAccount     → plain check only
#
#  The CALLER doesn't need to know WHICH account type it is.
#  It just calls withdraw() and the right behaviour runs.
 
# ── Three account classes with different withdraw() rules ───
 
class BankAccount:                           # basic account
 
    def __init__(self, owner, balance):
        self.owner     = owner
        self.__balance = balance
 
    def get_balance(self):
        return self.__balance
 
    def deposit(self, amount):
        self.__balance += amount
 
    def withdraw(self, amount):              # PARENT version
        if amount > self.__balance:
            print(f"  [{self.owner}] Not enough funds! "
                  f"Has ${self.__balance}, needs ${amount}")
            return
        self.__balance -= amount
        print(f"  [{self.owner}] Withdrew ${amount}. "
              f"Balance: ${self.__balance}")
 
 
class SavingsAccount(BankAccount):           # must keep $200 minimum
 
    def __init__(self, owner, balance):
        super().__init__(owner, balance)
        self.min_balance = 200               # cannot go below this
 
    # OVERRIDE withdraw() — savings-specific rule
    def withdraw(self, amount):
        remaining = self.get_balance() - amount
        if remaining < self.min_balance:
            print(f"  [{self.owner}] Blocked! Would fall below "
                  f"${self.min_balance} minimum. Balance stays at "
                  f"${self.get_balance()}")
            return
        # If the check passes, call the parent withdraw normally
        super().withdraw(amount)
 
 
class CurrentAccount(BankAccount):           # business account — allows overdraft
 
    def __init__(self, owner, balance):
        super().__init__(owner, balance)
        self.overdraft = 500                 # can go $500 below zero
 
    # OVERRIDE withdraw() — current-specific rule
    def withdraw(self, amount):
        if amount > self.get_balance() + self.overdraft:
            print(f"  [{self.owner}] Exceeds overdraft limit! "
                  f"Max you can take: ${self.get_balance() + self.overdraft}")
            return
        # Manually adjust since we're bypassing the parent's check
        # (we're allowed to go negative up to overdraft limit)
        new_bal = self.get_balance() - amount
        # Use deposit/withdraw trick to set the balance
        # easier: just call super and handle negative case
        super().deposit(self.overdraft)       # temporarily add overdraft buffer
        super().withdraw(amount + self.overdraft)  # withdraw + remove buffer
        print(f"  [{self.owner}] (overdraft used if balance went negative)")
 
 
# ── Simpler CurrentAccount using a protected balance directly ──
# (re-defining cleanly for clarity)
 
class CurrentAccount(BankAccount):
 
    def __init__(self, owner, balance):
        super().__init__(owner, balance)
        self.overdraft_limit = 500
 
    def withdraw(self, amount):
        max_allowed = self.get_balance() + self.overdraft_limit
        if amount > max_allowed:
            print(f"  [{self.owner}] Over limit! "
                  f"Max you can withdraw: ${max_allowed}")
            return
        # Directly update via deposit trick: remove balance, add overdraft room
        # Cleanest way at beginner level without exposing __balance:
        # We fake it by depositing the overdraft gap first, then withdrawing
        gap = amount - self.get_balance()
        if gap > 0:
            # Going into overdraft — deposit the gap first so parent check passes
            super().deposit(gap)
        super().withdraw(amount)
        if self.get_balance() < 0:
            print(f"  [{self.owner}] Warning: account is in overdraft!")
 
 
# ── THE MAGIC: one function, works on ALL account types ────────
#
#  This is the heart of polymorphism.
#  'account' could be ANY type — we don't check or care.
#  We just call .withdraw() and Python figures out which
#  version to run based on the actual object type.
 
def do_withdrawal(account, amount):
    print(f"\n  Trying to withdraw ${amount} from {account.owner}'s account...")
    account.withdraw(amount)    # ← same line, different behaviour each time!
 
 
# ── Try it: same function call, three different outcomes ───────
basic   = BankAccount("Charlie",  300)
savings = SavingsAccount("Alice", 500)
current = CurrentAccount("Bob",   300)
 
print("\n  -- Test 1: normal withdrawal everyone can afford --")
do_withdrawal(basic,   100)   # BankAccount.withdraw()   → simple check
do_withdrawal(savings, 100)   # SavingsAccount.withdraw()→ min balance check
do_withdrawal(current, 100)   # CurrentAccount.withdraw()→ overdraft check
 
print("\n  -- Test 2: withdrawal that would break savings minimum --")
do_withdrawal(savings, 400)   # blocked! would leave only $100 < $200 min
 
print("\n  -- Test 3: overdraft withdrawal (current account only) --")
do_withdrawal(current, 600)   # allowed! $300 balance + $500 overdraft = $800 max
 
print("\n  -- Test 4: everyone processes a list of accounts --")
accounts = [basic, savings, current]   # mixed list of different types!
for acc in accounts:
    do_withdrawal(acc, 50)    # same call on each — each behaves its own way
```

```python
# ────────────────────────────────────────────────────────────
# STEP 8 — __str__
# ────────────────────────────────────────────────────────────
#
#  When you do print(alice), Python calls alice.__str__()
#  to decide what text to show. Without it you get something
#  ugly like <__main__.BankAccount object at 0x7f3...>
#
#  Add __str__ to return a nice readable string instead.


print("\n" + "─"*45)
print("STEP 8 — __str__ (pretty printing)")
print("─"*45)

class BankAccount:

    def __init__(self, owner, balance):
        self.owner     = owner
        self.__balance = balance

    def get_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount <= 0:
            print("  Amount must be positive.")
            return
        self.__balance += amount
        print(f"  Deposited ${amount}. Balance: ${self.__balance}")

    def withdraw(self, amount):
        if amount <= 0:
            print("  Amount must be positive.")
            return
        if amount > self.__balance:
            print("  Not enough funds!")
            return
        self.__balance -= amount
        print(f"  Withdrew ${amount}. Balance: ${self.__balance}")

    # __str__ is called automatically by print()
    def __str__(self):
        return f"Account[{self.owner}] — Balance: ${self.__balance}"


class SavingsAccount(BankAccount):

    def __init__(self, owner, balance, interest_rate):
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        interest = round(self.get_balance() * self.interest_rate, 2)
        self.deposit(interest)

    # Override __str__ to show the interest rate too
    def __str__(self):
        return (
            f"SavingsAccount[{self.owner}] — "
            f"Balance: ${self.get_balance()} | "
            f"Rate: {self.interest_rate * 100}%"
        )


# ── Try it ───────────────────────────────────────────────────
acc     = BankAccount("Bob", 2000)
savings = SavingsAccount("Alice", 5000, 0.04)

print(acc)      # Account[Bob] — Balance: $2000
print(savings)  # SavingsAccount[Alice] — Balance: $5000 | Rate: 4.0%
```

```python
# ────────────────────────────────────────────────────────────
# FINAL DEMO — putting it all together
# ────────────────────────────────────────────────────────────

print("\n" + "="*45)
print("  FINAL DEMO — Full mini bank")
print("="*45 + "\n")

# Create two different account types
alice = SavingsAccount("Alice", 3000, 0.05)
bob   = BankAccount("Bob", 1500)

print("Starting positions:")
print(alice)
print(bob)

print("\nAlice saves more money and earns interest:")
alice.deposit(1000)
alice.add_interest()
print(alice)

print("\nBob deposits and withdraws:")
bob.deposit(500)
bob.withdraw(700)
bob.withdraw(9999)   # will fail gracefully
print(bob)

print("\nDone!")
```
